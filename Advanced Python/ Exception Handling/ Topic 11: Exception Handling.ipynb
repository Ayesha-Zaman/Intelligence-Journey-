### Exceptions

If things go wrong during the execution of the program(runtime). It generally happens when something unforeseen has happened.

- Exceptions are raised by python runtime
- You have to takle is on the fly

#### **Examples**

- Memory overflow
- Divide by 0 -> logical error
- Database error

```python
# Why is it important to handle exceptions
# -> User experience & Security
# how to handle exceptions
# -> Try except block
# Try: In try block we write the piece of code which can messmess up our program
# Except: In except we what to do after mess up.
```

```python
# let's create a file

with open('sample.txt', 'w') as f:
  f.write('Hello World')
```

```python
# try except demo

with open('sample.txt', 'r') as f:
  print(f.read())
```

```python
# let's suppose someone changed file name in server
# it's gonna give error
with open('sample1.txt', 'r') as f:
  print(f.read())
```

```python
# Handling through try & except

try:
  with open('sample1.txt', 'r') as f:
    print(f.read())
except:
  print('Sorry File not found')
```

```python
# Tip: whenever you are using external resources(could be a file, database, bluetooth connection) in your program always use try and except.
```

```python
# catching specific exception
try:
  n=5
  f = open('sample.txt', 'r')
  print(f.read())
  print(n)
  print(5/2)
  L = [1,2,3]
  L[100]
except FileNotFoundError:
  print('File not found')
except NameError:
  print('Variable not found')
except ZeroDivisionError:
  print('Cannot divide by 0')
except Exception as e:   # It's way to cath error which i haven't anticipated, always write it at the end
  print(e)
```

```python
# else
# Work flow: try -> except  OR  try -> else
try:
  f = open('sample1.txt','r')
except FileNotFoundError:
  print('File not found.')
except Exception:
  print('Something is wrong')
else:                           # code in else block will execute, when try block 100% works
  print(f.read())

```

```python
# finally
# Work flow: try -> except -> finally  OR  try -> else -> finally
try:
  f = open('sample1.txt','r')
except FileNotFoundError:
  print('File not found.')
except Exception:
  print('Something is wrong')
else:
  print(f.read())
finally:
  print('This is gonna print in any case.')
```

```python
# Benefit

class Bank:

  def __init__(self,balance):
    self.balance = balance

  def withdraw(self,amount):
    if amount < 0:
      raise Exception('Amount cannot be -ve')
    if self.balance < amount:
      raise Exception('You donot have enough balance')
    self.balance = self.balance - amount

obj = Bank(10000)
try:
  obj.withdraw(-5000)
except Exception as e:
  print(e)
else:
  print('Current balance: ',obj.balance)
```

```python
class MyException(Exception):
  def __init__(self,message):
    print(message)

class Bank:

  def __init__(self,balance):
    self.balance = balance

  def withdraw(self,amount):
    if amount < 0:
      raise MyException('Amount cannot be -ve')
    if self.balance < amount:
      raise MyException('You donot have enough balance')
    self.balance = self.balance - amount

obj = Bank(10000)
try:
  obj.withdraw(5000)
except MyException as e:
  pass
else:
  print('Current balance: ',obj.balance)
```

```python
# Simple Example
class SecurityError(Exception):

  def __init__(self,message):
    print(message)

  def logout(self):
    print('logout')

class Google:

  def __init__(self,name,email,password,device):
    self.name = name
    self.email = email
    self.password = password
    self.device = device

  def login(self,email,password,device):
    if device != self.device:
      raise SecurityError('Your account being accessed by another device')
    if email == self.email and password == self.password:
      print('welcome')
    else:
      print('login error')



obj = Google('ayesha','ayesha@gmail.com','1234','android')

try:
  obj.login('ayesha@gmail.com','1234','windows')
except SecurityError as e:
  e.logout()
else:
  print(obj.name)
finally:
  print('database connection closed')

```
