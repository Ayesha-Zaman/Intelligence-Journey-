### Namespaces

A namespace is a space that holds names(identifiers).Programmatically speaking, namespaces are dictionary of identifiers(keys) and their objects(values)

There are 4 types of namespaces:
- Builtin Namespace
- Global Namespace
- Enclosing Namespace
- Local Namespace

### Scope and LEGB Rule

A scope is a textual region of a Python program where a namespace is directly accessible.

The interpreter searches for a name from the inside out, looking in the local, enclosing, global, and finally the built-in scope. If the interpreter doesn’t find the name in any of these locations, then Python raises a NameError exception.

```python
# local and global
# global var
a = 2

def temp():
  # local var
  b = 3
  print(b)

temp()
print(a)
```

```python
# local and global -> editing global
# Error: we can use global variable in local scope but can't edit it. It's like we can read not write
a = 2

def temp():
  # local var
  a += 1
  print(a)

temp()
print(a)
```

```python
# we can't change it directly, but with the help of 'global' keyword
# we have to avoid it, beacuse not a good programming practice

a = 2

def temp():
  # local var
  global a
  a += 1
  print(a)

temp()
print(a)
```

```python
# Enclosing scope
# inner function -> local scope
# main program -> global function
# outer function -> enclose scope nonlocal
def outer():
  def inner():
    print('inner function')
  inner()
  print('outer function')


outer()
print('main program')
```

```python
# changing value from local to enclose -> use keyword 'nonlocal'
# nonlocal keyword
def outer():
  a = 1
  def inner():
    nonlocal a
    a += 1
    print('inner',a)
  inner()
  print('outer',a)


outer()
print('main program')
```

### Decorators

A decorator in python is a function that receives another function as input and adds some functionality(decoration) to it and returns it.

This can happen only because python functions are 1st class citizens.

There are 2 types of decorators available in python
- `Built in decorators` like `@staticmethod`, `@classmethod`, `@abstractmethod` and `@property` etc
- `User defined decorators` that we programmers can create according to our needs

```python
# functions are 1st class citicitizens
def func():
  print('Hello!')

a = func

a()
```

```python
# functions are 1st class citicitizens

def modify(func,num):
  return func(num)

def square(num):
  return num**2

modify(square,2)
```

```python
# simple example

def my_decorator(func):
  def wrapper():
    print('***********************')
    func()
    print('***********************')
  return wrapper

def hello():
  print('hello')


a = my_decorator(hello)
a()
```

```python
# simple example

def my_decorator(func):
  def wrapper():
    print('***********************')
    func()
    print('***********************')
  return wrapper

def hello():
  print('hello')

def display():
  print('hello ayesha')

a = my_decorator(hello)
a()

b = my_decorator(display)
b()
```

```python
# How decorators actually works? --> closure?
# decorator function based on clouser: that's mean even if the outer function get excuted and the scop got finish
# inner function still able to access the variables/attributes of outer function

# Example

def outer():
  a = 5
  def inner():
    print(a)
  return inner


b = outer()
b()
```

```python
# Better syntax?

def my_decorator(func):
  def wrapper():
    print('***********************')
    func()
    print('***********************')
  return wrapper

@my_decorator
def hello():
  print('hello')


hello()

```

```python
# anything meaningful?
# we created a function where we can find the execution time of any function

import time

def timer(func):
  def wrapper():
    start  = time.time()
    func()
    print("Time taken by", func.__name__, "is", time.time()-start, "sec.")
  return wrapper

@timer
def greeting():
  print("Assalam-O-Alikum Habibi!")

greeting()
```

```python
# Problem
# Does that work on all functions?
# Nope.. It will only work on those function which do not give any input

import time

def timer(func):
  def wrapper():
    start = time.time()
    func()
    print('time taken by',func.__name__,time.time()-start,'secs')
  return wrapper

@timer
def hello():
  print('hello wolrd')
  time.sleep(2)

@timer
def square(num):
  time.sleep(1)
  print(num**2)

square()
```

```python
# Solution --> *args
# it can handle functions without variable and with variables

# Problem
# Does that work on all functions?
# Nope.. It will only work on those function which do not give any input

import time

def timer(func):
  def wrapper(*args):
    start = time.time()
    func(*args)
    print('time taken by',func.__name__,time.time()-start,'secs')
  return wrapper

@timer
def hello():
  print('hello wolrd')
  time.sleep(2)

@timer
def square(num):
  time.sleep(1)
  print(num**2)

hello()
square(2)
```

```python
# Another Example
# Checking datatype

def sanity_check(data_type):
  def outer_wrapper(func):
    def inner_wrapper(*args):
      if type(*args) == data_type:
        func(*args)
      else:
        raise TypeError("This data type wouldn't work.")
    return inner_wrapper
  return outer_wrapper

@sanity_check(int)
def square(num):
  print(num**2)

@sanity_check(str)
def greet(name):
  print('hello',name)

square('two')
```
