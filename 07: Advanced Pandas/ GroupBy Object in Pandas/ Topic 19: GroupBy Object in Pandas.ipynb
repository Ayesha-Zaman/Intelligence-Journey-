```python
import pandas as pd
import numpy as np
```

```python
movies = pd.read_csv('/content/imdb-top-1000.csv')
```

```python
movies.head(4)
```

```python
# groupby works on categorical data
```

```python
genres = movies.groupby('Genre')
```

```python
# Applying builtin aggregation fuctions on groupby objects
# aggregation fuctions: max(), min(), sum(), std(), mean(), median()
genres.std(numeric_only=True)
```

```python
genres.max(numeric_only=True)
```

```python
# find the top 3 genres by total earning
genres['Gross'].sum().sort_values(ascending= False).head(3)
```

```python
# find the genre with highest avg IMDB rating
genres['IMDB_Rating'].mean().sort_values(ascending=False).head(1)
```

```python
# find director with most popularity
movies.groupby('Director')['No_of_Votes'].sum().sort_values(ascending=False).head(1)
```

```python
# find the highest rated movie of each genre
# movies.groupby('Genre')['IMDB_Rating'].max()
```

```python
# find number of movies done by each actor
# movies['Star1'].value_counts()
movies.groupby('Star1')['Series_Title'].count().sort_values(ascending=False)
```

```python
# GroupBy Attributes and Methods
# find total number of groups -> len
# find items in each group -> size
# first()/last() -> nth item
# get_group -> vs filtering
# groups
# describe
# sample
# nunique
```

```python
# find total number of groups -> len
# len(movies['Genre'].value_counts())
# movies['Genre'].nunique()
len(movies.groupby('Genre'))
```

```python
# find items in each group -> size
genres.size()
```

```python
# first()/last() -> nth item
# shows first item of each group/category
genres.first()
```

```python
# shows last item of each group/category
genres.last()
```

```python
# below code will show 3rd item of each group, if a group doesn't that much items it will drop them.
genres.nth(3)
```

```python
# get_group -> vs filtering
genres.get_group('Mystery')

movies[movies.Genre == 'Mystery']
```

```python
# groups: each group + store their value's index in list
genres.groups
```

```python
# describe
genres.describe()
```

```python
# sample select random row from each group
genres.sample()
```

```python
# wanna select 2 sample but group contain only one,but you also wanna display all groups
genres.sample(2, replace=True)
```

```python
# nunique: no. of unique values in each column of a group
genres.nunique()
```
