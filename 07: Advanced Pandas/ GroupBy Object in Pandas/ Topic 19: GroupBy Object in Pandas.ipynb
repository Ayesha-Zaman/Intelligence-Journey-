```python
import pandas as pd
import numpy as np
```

```python
movies = pd.read_csv('/content/imdb-top-1000.csv')
```

```python
# groupby works on categorical data
```

```python
genres = movies.groupby('Genre')
```

```python
# Applying builtin aggregation fuctions on groupby objects
# aggregation fuctions: max(), min(), sum(), std(), mean(), median()
genres.std(numeric_only=True)
```

```python
genres.max(numeric_only=True)
```

```python
# find the top 3 genres by total earning
genres['Gross'].sum().sort_values(ascending= False).head(3)
```

```python
# find the genre with highest avg IMDB rating
genres['IMDB_Rating'].mean().sort_values(ascending=False).head(1)
```

```python
# find total number of groups -> len
# len(movies['Genre'].value_counts())
# movies['Genre'].nunique()
len(movies.groupby('Genre'))
```

```python
# find items in each group -> size
genres.size()
```

```python
# shows last item of each group/category
genres.last()
```

```python
# below code will show 3rd item of each group, if a group doesn't that much items it will drop them.
genres.nth(3)
```

```python
# groups: each group + store their value's index in list
genres.groups
```

```python
# describe
genres.describe()
```

```python
# sample select random row from each group
genres.sample()
```

```python
# wanna select 2 sample but group contain only one,but you also wanna display all groups
genres.sample(2, replace=True)
```

```python
# nunique: no. of unique values in each column of a group
genres.nunique()
```

```python
# agg method
# passing dict
genres.agg(
    {
        'Runtime':'mean',
        'IMDB_Rating':'mean',
        'No_of_Votes':'sum',
        'Gross': 'sum',
        'Metascore': 'min'

    }
)
```

```python
# passing list
genres[movies.select_dtypes(include='number').columns].agg(['min', 'max', 'mean'])
```

```python
# Adding both the syntax
genres.agg(
    {
        'Runtime':['min','mean'],
        'IMDB_Rating':'mean',
        'No_of_Votes':['sum','max'],
        'Gross': 'sum',
        'Metascore': 'min'

    }
)
```

```python
# looping on groups
# it gives 2 values group and data in strings and data frame

for group,data in genres:
  print(data)
```

```python
# find the highest rated movie of each genre
df = pd.DataFrame(columns=movies.columns)
temp_df = []
for group,data in genres:
  top_movies = data[data['IMDB_Rating'] == data['IMDB_Rating'].max()]
  temp_df.append(top_movies)

df = pd.concat(temp_df)
df
```

```python
# split (apply) combine    <- aply strategy
# apply -> builtin function

genres.apply('min')
```

```python
# find number of movies starting with A for each group
for group,data in genres:
  A_movies = data['Series_Title'].str.startswith('A').sum()
  print(group,A_movies)
```

```python
# find number of movies starting with A for each group
def A_movies(group):
  return group['Series_Title'].str.startswith('A').sum()
```

```python
genres.apply(A_movies)
```

```python
# find ranking of each movie in the group according to IMDB score
def Rank(group):
  group['genre_rank'] = group['IMDB_Rating'].rank(ascending=False, method = "dense")
  return group
```

```python
genres.apply(Rank)
```

```python
# find normalized IMDB rating group wise

def normal(group):
  group['norm_rating'] = (group['IMDB_Rating'] - group['IMDB_Rating'].min())/(group['IMDB_Rating'].max() - group['IMDB_Rating'].min())
  return group

genres.apply(normal)
```

```python
ipl = pd.read_csv('/content/deliveries.csv')
ipl.head()
```

```python
ipl.shape
```

```python
# find the top 10 batsman in terms of runs
ipl.groupby('batsman')['batsman_runs'].sum().sort_values(ascending=False).head(10)
```

```python
# find the batsman with max no of sixes
ipl[ipl['batsman_runs'] == 6].groupby('batsman')['batsman_runs'].value_counts().sort_values(ascending=False).head(1).index[0]
```

```python
# find batsman with most number of 4's and 6's in last 5 overs
temp_df = ipl[ipl['over'] > 15]
temp_df = temp_df[(temp_df['batsman_runs'] == 4) | (temp_df['batsman_runs'] == 6)]
temp_df.groupby('batsman')['batsman'].value_counts().sort_values(ascending=False).head(1).index[0]
```

```python
# find V Kohli's record against all teams
ipl[ipl['batsman'] == 'V Kohli'].groupby('bowling_team')['batsman_runs'].sum().reset_index().sort_values('batsman_runs',ascending=False)
```

```python
# Create a function that can return the highest score of any batsman
temp_df = ipl[ipl['batsman'] == 'V Kohli']
temp_df.groupby('match_id')['batsman_runs'].sum().sort_values(ascending=False).head(1).values[0]
```

```python
def highest(batsman):
  temp_df = ipl[ipl['batsman'] == batsman]
  return temp_df.groupby('match_id')['batsman_runs'].sum().sort_values(ascending=False).head(1).values[0]
```

```python
highest('DA Warner')
```
