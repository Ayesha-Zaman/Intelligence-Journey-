```python
# import titanic
df = pd.read_csv('/content/titanic.csv')
df['Name']
```

```python
# Common Functions
# lower/upper/capitalize/title
df['Name'].str.upper()
df['Name'].str.capitalize()
df['Name'].str.title()
# len
df['Name'][df['Name'].str.len() == 82]
# strip
"                   ayesha                              ".strip()
df['Name'].str.strip()
```

```python
# split -> get
# adding lastname column from extracting from Name coulmn
df['lastname']= df['Name'].str.split(',').str.get(0)
```

```python
df.head()
```

```python
# now extracting title and first name
df[['title','firstname']] = df['Name'].str.split(',').str.get(1).str.strip().str.split(' ', n=1, expand=True)
df.head()
```

```python
df['title'].value_counts()
```

```python
# replace
# combining Miss. Ms. & Mlle. because they all are same

df['title'] = df['title'].str.replace('Ms.', 'Miss.')
df['title'] = df['title'].str.replace('Mlle.', 'Miss.')
```

```python
df['title'].value_counts()
```

```python
# filtering
# startswith/endswith
df[df['firstname'].str.endswith('A')]
# isdigit/isalpha...
df[df['firstname'].str.isdigit()]
```

```python
# applying regex
# contains
# search john -> both case
df[df['firstname'].str.contains('john', case=False)]
# find lastname with start and end char vowel
df[df['lastname'].str.contains('^[aeiouAEIOU].+[aeiouAEIOU]$')]
```
