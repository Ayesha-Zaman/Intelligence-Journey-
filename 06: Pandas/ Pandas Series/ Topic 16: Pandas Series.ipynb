### Pandas Series

A Pandas Series is like a column in a table. It is a 1-D array holding data of any type.

### Importing Pandas

```python
import numpy as np
import pandas as pd
```

```python
# like python, pandas also have many objects. The two major or prominent are:
# Series
# DataFrame
```

### Series from lists

```python
# string
countries = ['Pakistan','Bangladesh','America', 'India', 'China']
pd.Series(countries)
```

```python
# In pandas 99% of the time "dtype: object" refers to string datatype
```

```python
# Integers
runs = [31,45,67,27,100]
run_ser = pd.Series(runs)
run_ser
```

```python
# custom index
marks = [67,57,89,100]
subjects = ['maths','english','science','Quranic_std']

pd.Series(marks, index=subjects)
```

```python
# setting a name
marks = pd.Series(marks, index=subjects, name="Ayesha's school marks")
marks
```

### Series from dict

```python
marks = {
    'maths':67,
    'english':57,
    'science':89,
    'Quranic_std':100
}

marks_series = pd.Series(marks, name="Ayesha's school marks")
marks_series
```

```python
# size      ->no. of items
marks_series.size
```

```python
# dtype
marks_series.dtype
```

```python
# name
marks_series.name
```

```python
# is_unique   ->are all items unique?
marks_series.is_unique
```

```python
pd.Series([1,1,2,3,4,3,5]).is_unique
```

```python
# index
marks_series.index
```

```python
run_ser.index
```

```python
# value
marks_series.values
```
