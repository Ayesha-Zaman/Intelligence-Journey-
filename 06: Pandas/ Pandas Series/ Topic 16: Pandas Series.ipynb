### Pandas Series

A Pandas Series is like a column in a table. It is a 1-D array holding data of any type.

### Series from lists

```python
# string
countries = ['Pakistan','Bangladesh','America', 'India', 'China']
pd.Series(countries)
```

```python
# In pandas 99% of the time "dtype: object" refers to string datatype
```

```python
# Integers
runs = [31,45,67,27,100]
run_ser = pd.Series(runs)
run_ser
```

```python
# custom index
marks = [67,57,89,100]
subjects = ['maths','english','science','Quranic_std']

pd.Series(marks, index=subjects)
```

```python
# setting a name
marks = pd.Series(marks, index=subjects, name="Ayesha's school marks")
marks
```
