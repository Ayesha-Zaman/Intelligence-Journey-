###`Q-4.`Write a function to change the data type of given a column or a Series. Function takes series and data type as input, returns the converted series.
```
series = pd.Series([1,2,'Python', 2.0, True, 100])
change to float type data
```
Note: Read about pd.to_numeric()

````python
def change_dtype(series):
  return pd.to_numeric(series, errors='coerce')


series = pd.Series([1,2,'Python', 2.0, True, 100])
change_dtype(series)
````

````python
# i.
items = pd.read_csv('/content/items.csv', index_col=['item_name']).squeeze("columns")
````

````python
# ii(2).
def rupees(x):
  try:
    y = x[1:]
  except:
    y = x
  return float(y)*279.90
items.apply(rupees)
````

````python
# iii.
items = items.apply(rupees)
````
