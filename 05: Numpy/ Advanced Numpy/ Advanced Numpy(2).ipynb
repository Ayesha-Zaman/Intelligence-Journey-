```python
import numpy as np
```

```python
# plotting 3D graphs
# meshgrids
a = np.linspace(-10,9,20)
b = np.linspace(-10,9,20)

print(a,b)
```

```python
# Notes: meshgrids are used to make coordinates which are further feeded into a function to generate 3D plots.
# utility: it's helps to plot 3D graphs
```

```python
xx, yy = np.meshgrid(a,b)
xx, yy
```

```python
import matplotlib.pyplot as plt
plt.scatter(xx, yy)
```

```python
# parabola function

def func(x,y):
  return (x)**2 + (y)**2

zz = func(xx,yy)
zz
```

```python
# another function, we can change equation
def func(x,y):
  return 2*(np.sin(x)) + (y)*2

zz = func(xx,yy)
```

```python
import plotly.express as px
import plotly.graph_objects as go

fig = px.scatter_3d()

fig.add_trace(go.Surface(x=xx,y=yy,z=zz))

fig.show()
```
